**3. Discussions:**

A data dashboard is a visual interface that displays key information and performance indicators in a concise and interactive manner. Dashboards consolidate multiple data sources to provide users with insights for decision-making and performance tracking.

Principles of Effective Dashboards:

Clarity and Simplicity: Visuals should be easy to read and interpret.

Real-Time or Dynamic Data: Dashboards often pull updated data automatically.

Interactivity: Users should be able to filter, sort, or drill down into specific metrics.

Action-Oriented Insights: The dashboard should guide users toward decisions or next steps.

Taxonomies of Data Dashboards:

Data Updates – Dashboards can show static (snapshot) or dynamic (real-time) data.

User Interaction – Some dashboards allow filtering, searching, and selecting parameters.

Organizational Function – Dashboards can serve strategic, analytical, or operational functions.

For example:

Strategic dashboards help top management monitor KPIs.

Analytical dashboards help analysts explore data trends.

Operational dashboards support real-time monitoring of systems and workflows.

```python
import pandas as pd
import numpy as np
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
```

```python
# Simulated company performance data
np.random.seed(42)
months = pd.date_range("2024-01-01", periods=12, freq="M")
regions = ["North", "South", "East", "West"]
products = ["Laptop", "Tablet", "Phone"]

data = {
    "Month": np.random.choice(months, 200),
    "Region": np.random.choice(regions, 200),
    "Product": np.random.choice(products, 200),
    "Sales": np.random.randint(5000, 20000, 200),
    "Profit": np.random.randint(500, 5000, 200)
}

df = pd.DataFrame(data)
df.head()
```

```python
# Aggregated views
sales_by_region = df.groupby("Region", as_index=False)["Sales"].sum()
sales_by_product = df.groupby("Product", as_index=False)["Sales"].sum()
sales_trend = df.groupby("Month", as_index=False)["Sales"].sum().sort_values("Month")

# Create dashboard layout
fig = make_subplots(rows=2, cols=2,
                    subplot_titles=("Sales by Region", "Sales by Product", "Monthly Sales Trend", "Profit vs Sales"),
                    specs=[[{"type": "bar"}, {"type": "pie"}],
                           [{"colspan": 2}, None]])

# Add charts
fig.add_trace(go.Bar(x=sales_by_region["Region"], y=sales_by_region["Sales"], name="Region Sales"), row=1, col=1)
fig.add_trace(go.Pie(labels=sales_by_product["Product"], values=sales_by_product["Sales"], name="Product Sales"), row=1, col=2)
fig.add_trace(go.Scatter(x=sales_trend["Month"], y=sales_trend["Sales"], mode="lines+markers", name="Monthly Trend"), row=2, col=1)

fig.update_layout(height=700, width=950, title_text="📊 Interactive Data Dashboard Example")
fig.show()
```

```python
# Interactive dashboard: filterable by product
fig = px.bar(df, x="Region", y="Sales", color="Product", barmode="group",
             title="Interactive Dashboard: Sales by Region and Product",
             hover_data=["Profit"])
fig.show()
```

```python
import pandas as pd
import numpy as np
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Simulated store sales data
np.random.seed(1)
months = pd.date_range("2024-01-01", periods=12, freq="ME")
branches = ["Manila", "Cebu", "Davao", "Baguio"]
categories = ["Food", "Drinks", "Snacks"]

data = {
    "Month": np.random.choice(months, 200),
    "Branch": np.random.choice(branches, 200),
    "Category": np.random.choice(categories, 200),
    "Sales": np.random.randint(5000, 20000, 200),
    "Profit": np.random.randint(500, 5000, 200)
}

df = pd.DataFrame(data)
df.head()
```

```python
# Aggregated views
sales_by_branch = df.groupby("Branch", as_index=False)["Sales"].sum()
sales_by_category = df.groupby("Category", as_index=False)["Sales"].sum()
sales_trend = df.groupby("Month", as_index=False)["Sales"].sum().sort_values("Month")

# Create dashboard layout
fig = make_subplots(rows=2, cols=2,
                    subplot_titles=("Sales by Branch", "Sales by Category", "Monthly Sales Trend"),
                    specs=[[{"type": "bar"}, {"type": "pie"}],
                           [{"colspan": 2}, None]])

# Add charts
fig.add_trace(go.Bar(x=sales_by_branch["Branch"], y=sales_by_branch["Sales"], name="Branch Sales"), row=1, col=1)
fig.add_trace(go.Pie(labels=sales_by_category["Category"], values=sales_by_category["Sales"], name="Category Sales"), row=1, col=2)
fig.add_trace(go.Scatter(x=sales_trend["Month"], y=sales_trend["Sales"], mode="lines+markers", name="Monthly Trend"), row=2, col=1)

# Labels for readability
fig.update_xaxes(title_text="Branch", row=1, col=1)
fig.update_yaxes(title_text="Total Sales (₱)", row=1, col=1)
fig.update_xaxes(title_text="Month", row=2, col=1)
fig.update_yaxes(title_text="Total Sales (₱)", row=2, col=1)

fig.update_layout(height=700, width=950, title_text="🏪 Store Sales Dashboard")
fig.show()
```

**Task 2:** Data Dashboard Taxonomies

Add interactivity features such as:

A dropdown menu to select regions or products

A dynamic update (simulating real-time data) using Python’s update functions

Explain how your dashboard fits into:

Data Update Type (Static, Scheduled, or Real-Time)

User Interaction (Filter, Drilldown, etc.)

Organizational Function (Strategic, Analytical, Operational)

```python
# Interactive feature 1: Dropdown menu to select a branch
fig = go.Figure()

for b in branches:
    branch_data = df[df["Branch"] == b].groupby("Category", as_index=False)["Sales"].sum()
    fig.add_trace(go.Bar(x=branch_data["Category"], y=branch_data["Sales"],
                         name=b, visible=(b == branches[0])))

# Dropdown buttons
buttons = []
for i, b in enumerate(branches):
    visible = [False] * len(branches)
    visible[i] = True
    buttons.append(dict(label=b, method="update",
                        args=[{"visible": visible},
                              {"title": f"Sales by Category - {b} Branch"}]))

fig.update_layout(
    updatemenus=[dict(buttons=buttons, direction="down", x=0.1, y=1.15)],
    title=f"Sales by Category - {branches[0]} Branch",
    xaxis_title="Category", yaxis_title="Total Sales (₱)"
)
fig.show()
```

```python
# Interactive feature 2: Dynamic update (simulating real-time data)
def refresh_dashboard(df):
    # Add 10 new random sales records (like new data arriving)
    new_data = pd.DataFrame({
        "Month": np.random.choice(months, 10),
        "Branch": np.random.choice(branches, 10),
        "Category": np.random.choice(categories, 10),
        "Sales": np.random.randint(5000, 20000, 10),
        "Profit": np.random.randint(500, 5000, 10)
    })
    df = pd.concat([df, new_data], ignore_index=True)

    # Redraw the chart with the updated data
    updated = df.groupby("Branch", as_index=False)["Sales"].sum()
    fig = px.bar(updated, x="Branch", y="Sales", color="Branch",
                 title=f"Updated Sales by Branch (Total records: {len(df)})")
    fig.show()
    return df

# Run the update 2 times
df = refresh_dashboard(df)
df = refresh_dashboard(df)
```

**Explanation**

Data Update Type: Real-Time (simulated). The refresh_dashboard() function adds new data and redraws the chart, like a live dashboard. A real one would pull from a database or sensor.<br>

User Interaction: Filter. The dropdown lets the user pick a branch. Hovering over the charts shows exact values.<br>

Organizational Function: Analytical. It lets managers compare branches and categories to find which ones sell best.

**5. Supplementary Activity:**

Create a mini dashboard project that tracks student grades or attendance:

1. Include 3 visualizations (e.g., average grades by subject, attendance by week, and performance trend).

2. Add at least one interactive feature (dropdown or slider).

3. Explain how your dashboard helps an instructor or department make data-driven decisions.

```python
#Simulated student data
np.random.seed(5)
weeks = list(range(1, 11))
subjects = ["Math", "Science", "English", "Programming", "Data Analytics"]

df = pd.DataFrame({
    "Week": np.tile(weeks, len(subjects)),
    "Subject": np.repeat(subjects, len(weeks)),
    "Grade": np.random.randint(70, 100, 50),
    "Attendance (%)": np.random.randint(75, 100, 50)
})

#Aggregated data
avg_grades = df.groupby("Subject", as_index=False)["Grade"].mean()
attendance_week = df.groupby("Week", as_index=False)["Attendance (%)"].mean()

#Dashboard layout
fig = make_subplots(rows=2, cols=2,
                    subplot_titles=("Average Grade by Subject", "Attendance by Week", "Performance Trend per Subject"),
                    specs=[[{"type": "bar"}, {"type": "scatter"}],
                           [{"colspan": 2}, None]])

#Chart 1: Average grade by subject
fig.add_trace(go.Bar(x=avg_grades["Subject"], y=avg_grades["Grade"],
                     name="Average Grade", showlegend=False), row=1, col=1)

#Chart 2: Attendance by week
fig.add_trace(go.Scatter(x=attendance_week["Week"], y=attendance_week["Attendance (%)"],
                         mode="lines+markers", name="Attendance", showlegend=False), row=1, col=2)

#Chart 3: Performance trend (one line per subject)
for s in subjects:
    sub = df[df["Subject"] == s]
    fig.add_trace(go.Scatter(x=sub["Week"], y=sub["Grade"], mode="lines+markers",
                             name=s, visible=(s == subjects[0])), row=2, col=1)

#Interactive feature: dropdown to choose subject for the trend chart
buttons = []
for i, s in enumerate(subjects):
    visible = [True, True] + [j == i for j in range(len(subjects))]
    buttons.append(dict(label=s, method="update", args=[{"visible": visible}]))

buttons.append(dict(label="All Subjects", method="update",
                    args=[{"visible": [True, True] + [True] * len(subjects)}]))

fig.update_layout(
    updatemenus=[dict(buttons=buttons, direction="down", x=0.1, y=0.42)],
    height=700, width=950, title_text="🎓 Student Grades and Attendance Dashboard"
)
fig.update_xaxes(title_text="Week", row=1, col=2)
fig.update_xaxes(title_text="Week", row=2, col=1)
fig.update_yaxes(title_text="Grade", row=2, col=1)
fig.show()
```

**How the dashboard helps an instructor or department:**<br>

The bar chart shows which subjects have low average grades, so the department knows where students need more help. The attendance chart shows which weeks had low attendance, so the instructor can look for the cause, such as exam weeks. The trend chart, filtered with the dropdown, shows whether a subject is improving or getting worse. With this, instructors can act early (extra lessons, reminders) based on data instead of guessing.
