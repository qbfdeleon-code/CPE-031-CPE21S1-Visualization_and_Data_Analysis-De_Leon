**4. Procedures:**

Step 1. Import Libraries and Create Sample Data

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Sample sales dataset
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [200, 250, 180, 300, 280, 350],
    "Marketing_Spend": [50, 60, 45, 70, 65, 80]
}
df = pd.DataFrame(data)
df
```

Step 2. Know Your Audience — Compare Chart Styles

```python
# Executive audience: prefer clarity (simple bar)
plt.figure(figsize=(7,4))
sns.barplot(x="Month", y="Sales", data=df, color="steelblue")
plt.title("Executive View: Monthly Sales Overview", fontsize=13)
plt.show()

# Analyst audience: prefer detailed trends (line chart)
plt.figure(figsize=(7,4))
sns.lineplot(x="Month", y="Sales", data=df, marker="o", label="Sales")
sns.lineplot(x="Month", y="Marketing_Spend", data=df, marker="o", label="Marketing Spend")
plt.title("Analyst View: Sales vs Marketing Trend", fontsize=13)
plt.legend()
plt.show()
```

**Task 1:** Create one additional visualization suited for a different audience (e.g., a customer, investor, or teammate).

```python

plt.figure(figsize=(7,4))
sns.barplot(x="Month", y="Sales", data=df, color="steelblue")
plt.title("Customer View", fontsize=13)
plt.ylabel("Sales")
plt.show()

# Investor audience: return on marketing
df["Sales_per_Marketing"] = df["Sales"] / df["Marketing_Spend"]

plt.figure(figsize=(7,4))
sns.lineplot(x="Month", y="Sales_per_Marketing", data=df, marker="o", color= "crimson")
plt.title("Investor view", fontsize=13)
plt.ylabel("Sales / Marketing Spend")
plt.show()
```

Step 3. Know Your Message — Highlight What Matters

```python
# Highlight the highest sales month (preattentive color cue)
highlight = df["Sales"].idxmax()

colors = ["gray" if i != highlight else "crimson" for i in range(len(df))]
plt.figure(figsize=(7,4))
plt.bar(df["Month"], df["Sales"], color=colors)
plt.title("Highlighting the Best Month in Sales", fontsize=13)
plt.show()
```

**Task 2:** Use color or annotations to emphasize a key insight (e.g., lowest point, sudden increase).

```python
# Highlight the highest sales month
lowest = df["Sales"].idxmin()

colors = ["gray" if i != lowest else "crimson" for i in range(len(df))]
plt.figure(figsize=(7,4))
plt.bar(df["Month"], df["Sales"], color=colors)

plt.annotate("Lowest sales (180)\nMarketing was also lowest (45)",
             xy=(lowest, df["Sales"][lowest]),
             xytext=(lowest + 1, df["Sales"][lowest] + 90),
             arrowprops=dict(arrowstyle="->", color="crimson"),
             fontsize=10, color="crimson")

plt.title("March Was the Weakest Month", fontsize=13)
plt.ylabel("Sales")
plt.show()
```

Step 4. Storytelling with Charts — Beginning, Middle, End

```python
# Beginning: Context
plt.figure(figsize=(7,4))
sns.barplot(x="Month", y="Sales", data=df, color="skyblue")
plt.title("Beginning: Sales Started Steady", fontsize=13)
plt.show()

# Middle: Change or Challenge
plt.figure(figsize=(7,4))
sns.lineplot(x="Month", y="Sales", data=df, color="orange", marker="o")
plt.title("Middle: A Drop in March Due to Low Marketing", fontsize=13)
plt.show()

# End: Resolution
plt.figure(figsize=(7,4))
sns.barplot(x="Month", y="Sales", data=df, color="limegreen")
plt.title("End: Recovery and Growth After March", fontsize=13)
plt.show()
```

**Task 3:** Write one or two lines describing your data story from beginning → middle → end.

**Answer:** Sales started steady in January and February (Beginning). In March, sales dropped to 180 when marketing spend was also at its lowest (Middle). After marketing spend increased again, sales recovered and reached a peak of 350 in June (End), showing that higher marketing effort goes together with higher sales.

Step 5. Bringing It All Together — Presentation Design

```python
# Simple combined story chart
plt.figure(figsize=(8,5))
sns.lineplot(x="Month", y="Sales", data=df, marker="o", linewidth=2, color="royalblue", label="Sales")
sns.lineplot(x="Month", y="Marketing_Spend", data=df, marker="o", linestyle="--", color="orange", label="Marketing Spend")

plt.fill_between(df["Month"], df["Sales"], alpha=0.1, color="royalblue")
plt.title("From Effort to Impact: Marketing Spend Drives Sales Growth", fontsize=13)
plt.legend()
plt.show()
```

**Supplementary Activity: Sleep Health and Lifestyle Dataset (Kaggle)**<br>
***Dataset: https://www.kaggle.com/datasets/uom190346a/sleep-health-and-lifestyle-dataset***


```python
sleep_df = pd.read_csv("/content/HoA10p1_Sleep_health_and_lifestyle_dataset.csv")
sleep_df.head()
```

```python

#Average sleep quality for each occupation
avg_sleep = sleep_df.groupby("Occupation")["Quality of Sleep"].mean().sort_values().reset_index()
avg_sleep

#Highlight the lowest value using a distinct color
lowest = avg_sleep["Quality of Sleep"].idxmin()
colors = ["lightgray" if i != lowest else "crimson" for i in range(len(avg_sleep))]

plt.figure(figsize=(9,5))
plt.bar(avg_sleep["Occupation"], avg_sleep["Quality of Sleep"], color=colors)

#Annotation that tells the story behind the data
worst_job = avg_sleep["Occupation"][lowest]
worst_score = avg_sleep["Quality of Sleep"][lowest]
plt.annotate(f"{worst_job} has the\nlowest sleep quality ({worst_score:.1f}/10)",
             xy=(lowest, worst_score),
             xytext=(lowest + 1.5, worst_score + 0.6),
             arrowprops=dict(arrowstyle="->", color="crimson"),
             fontsize=10, color="crimson")
plt.title("Which Occupation Sleeps the Worst? Average Sleep Quality by Occupation", fontsize=13)
plt.xlabel("Occupation")
plt.ylabel("Average Quality of Sleep (1-10)")
plt.xticks(rotation=45, ha="right")
plt.ylim(0, 10)
plt.tight_layout()
plt.show()
```

**6. Conclusion/Learnings/Analysis:**

* **Learnings**:
different audiences need different visuals: executives want simple bar charts, while analysts prefer detailed line charts. Adding colors and labels helps people spot key takeaways immediately, like the highest or lowest points, without having to search. Framing data with a clear beginning, middle, and end turns plain numbers into a convincing story that guides better decisions.

* **Analysis**:
Sales Representatives have the worst sleep quality at 4.0 out of 10, while Engineers get the best sleep at 8.4 out of 10. Overall, jobs with steady office routines like engineers, lawyers, and accountants report much better sleep than high-stress or sales roles.
