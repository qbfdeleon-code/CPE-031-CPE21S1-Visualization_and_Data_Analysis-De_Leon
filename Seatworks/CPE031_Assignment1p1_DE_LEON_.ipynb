```python
# 1. Load the dataset and show the first 5 and last 5 datapoints
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/vda/youtube_subscribers_data.csv")
```

```python
df.isna().sum() #first 5 data
```

```python
# 2. Check if there are any missing values in the dataset.
df.isna().sum()
```

```python
# 3. Select only the required columns and rename 'Subscribers (millions)' to 'Subscribers'
df = df[['Name', 'Subscribers (millions)', 'Category', 'Country']]
df.head()
```

```python
# 4. Convert to UNKOWN to NaN
df['Subscribers (millions)'] = pd.to_numeric(df['Subscribers (millions)'], errors='coerce')
df = df.dropna() # remove the NaNs
df['Subscribers (millions)'] = df['Subscribers (millions)'].astype(int) # convert to int
df = df.rename(columns={'Subscribers (millions)': 'Subscribers'}) #rename subsribers
df.head()
```
